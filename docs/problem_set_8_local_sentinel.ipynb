```python
import json
import os
from pathlib import Path

credentials_path = Path("kaggle.json")

if not credentials_path.exists():
    raise FileNotFoundError(
        "kaggle.json was not found. Place it in the same folder as the notebook."
    )

with credentials_path.open("r", encoding="utf-8") as file:
    credentials = json.load(file)

os.environ["KAGGLE_USERNAME"] = credentials["username"]
os.environ["KAGGLE_KEY"] = credentials["key"]

print("Kaggle credentials configured successfully.")
```

```python
assert os.getenv("KAGGLE_USERNAME")
assert os.getenv("KAGGLE_KEY")

print("KAGGLE_USERNAME and KAGGLE_KEY are ready.")
```

```python
import kagglehub

MODEL_HANDLE = "google/gemma-4/transformers/gemma-4-e2b-it"

model_path = kagglehub.model_download(MODEL_HANDLE)

print("Model downloaded successfully.")
print("Model path:", model_path)
```

```python
from transformers import pipeline

sentinel = pipeline(
    task="image-text-to-text",
    model=model_path,
    device_map="auto",
    dtype="auto"
)

print("Inference pipeline initialized successfully.")
```

```python
question = (
    "Why is memory footprint reduction critical when running "
    "parallel physical simulations on the same hardware? "
    "Answer in two concise sentences."
)

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": question,
            }
        ],
    }
]

output = sentinel(
    text=messages,
    return_full_text=False,
    generate_kwargs={
        "max_new_tokens": 50,
        "do_sample": False,
    },
)

print(output[0]["generated_text"])
```

Question: Why is this memory footprint reduction a critical factor when executing parallel physical simulations on the same hardware?

Answer: Reducing memory footprint is critical to prevent resource contention and thrashing when multiple parallel simulations compete for limited physical memory. This ensures each simulation has sufficient dedicated space, leading to stable performance and efficient utilization of the shared hardware.
